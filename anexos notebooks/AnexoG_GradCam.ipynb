```python
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
import cv2

from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing import image
from tensorflow.keras.applications.resnet50 import preprocess_input
```

```python
model = load_model("modelo1_resnet50_final.h5")

#Definimos las clases
class_labels = ['AgeDegeneration','Myopia','cataract','diabetic_retinopathy','glaucoma','healthy']
```

```python
img_path = "dataset_final/test/Myopia/myopia_0072.jpg"

img = image.load_img(img_path, target_size=(224,224))
img_array = image.img_to_array(img)

img_array = np.expand_dims(img_array, axis=0)
img_array = preprocess_input(img_array)
```

```python
preds = model.predict(img_array)

predicted_class = class_labels[np.argmax(preds)]
print("Predicción:", predicted_class)
```

```python
last_conv_layer_name = "conv5_block3_out"
```

```python
def get_gradcam_heatmap(img_array, model, last_conv_layer_name, pred_index=None):

    grad_model = tf.keras.models.Model(
        [model.inputs],
        [model.get_layer(last_conv_layer_name).output, model.output]
    )

    with tf.GradientTape() as tape:
        conv_outputs, predictions = grad_model(img_array)

        if isinstance(predictions, list):
            predictions = predictions[0]

        if pred_index is None:
            pred_index = tf.argmax(predictions[0])

        class_channel = predictions[:, pred_index]

    grads = tape.gradient(class_channel, conv_outputs)

    pooled_grads = tf.reduce_mean(grads, axis=(0, 1, 2))

    conv_outputs = conv_outputs[0]

    heatmap = conv_outputs @ pooled_grads[..., tf.newaxis]
    heatmap = tf.squeeze(heatmap)

    heatmap = tf.maximum(heatmap, 0) / tf.math.reduce_max(heatmap)

    return heatmap.numpy()
```

```python
heatmap = get_gradcam_heatmap(img_array, model, last_conv_layer_name)
```

```python
plt.imshow(heatmap, cmap='jet')
plt.title("Grad-CAM Heatmap")
plt.axis('off')
plt.show()
```

```python
# Leer imagen original con OpenCV
img = cv2.imread(img_path)
img = cv2.resize(img, (224,224))

# Convertir heatmap
heatmap = np.uint8(255 * heatmap)
heatmap = cv2.resize(heatmap, (224,224))
heatmap = cv2.applyColorMap(heatmap, cv2.COLORMAP_JET)

# Superposición correcta
superimposed_img = cv2.addWeighted(img, 0.6, heatmap, 0.4, 0)

# Mostrar
plt.figure(figsize=(6,6))
plt.imshow(cv2.cvtColor(superimposed_img, cv2.COLOR_BGR2RGB))
plt.title(f"Grad-CAM: {predicted_class}")
plt.axis('off')
plt.show()
```
